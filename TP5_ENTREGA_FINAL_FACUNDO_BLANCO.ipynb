##Incongruencias

```python
#Incongruencias a revisar

#Superficie cubierta > superficie total
mask_sup = df_tp_clean["surface_covered"] > df_tp_clean["surface_total"]
print("Registros con covered > total:", mask_sup.sum())
print(df_tp_clean.loc[mask_sup, ["surface_total","surface_covered","rooms","bathrooms","price"]].head(60))
```

```python
df_incoh = df_tp_clean[df_tp_clean["surface_covered"] > df_tp_clean["surface_total"]].copy()

# Definir máscara de diferencia "pequeña" (<=20% de la superficie cubierta)
mask_swap = (df_incoh["surface_total"] <= df_incoh["surface_covered"]*1.2)

# Intercambio donde parece un swap
tmp_total = df_incoh.loc[mask_swap, "surface_total"].copy()
df_tp_clean.loc[df_incoh.index[mask_swap], "surface_total"] = df_incoh.loc[mask_swap, "surface_covered"]
df_tp_clean.loc[df_incoh.index[mask_swap], "surface_covered"] = tmp_total

# Donde la diferencia es grande → setear covered como NaN
mask_big_diff = ~mask_swap
df_tp_clean.loc[df_incoh.index[mask_big_diff], "surface_covered"] = np.nan

print("Intercambios realizados:", mask_swap.sum())
print("Marcados como NaN:", mask_big_diff.sum())
```

```python
# bedrooms vs rooms
# Detección de registros incoherentes
mask_bedrooms = (
    ((df_tp_clean["rooms"] == 1) & (df_tp_clean["bedrooms"] > 0)) |
    ((df_tp_clean["rooms"] == 2) & (df_tp_clean["bedrooms"] > 1)) |
    ((df_tp_clean["rooms"] == 3) & (df_tp_clean["bedrooms"] > 2))
)

print("Registros con bedrooms > lógico:", mask_bedrooms.sum())

# Mostrar algunos
print(df_tp_clean.loc[mask_bedrooms, ["rooms", "bedrooms", "surface_total", "price"]].head())

# NaN para imputar más adelante
df_tp_clean.loc[mask_bedrooms, "bedrooms"] = np.nan

print("Quedan nulos en bedrooms:", df_tp_clean["bedrooms"].isna().sum())
```

##Outliers en los barrios con Z-score

```python
outliers_local_media_std = []

for barrio, subset in df_tp_clean.groupby("l3")["price_usd"]:
    mean = subset.mean()
    std = subset.std()

    lower = mean - 2*std
    upper = mean + 2*std

    outlier_mask = (subset < lower) | (subset > upper)
    prop_outliers = outlier_mask.mean()

    outliers_local_media_std.append({
        "l3": barrio,
        "mean": mean,
        "std": std,
        "lower_limit": lower,
        "upper_limit": upper,
        "prop_outliers_local": prop_outliers
    })

outliers_local_media_std = (
    pd.DataFrame(outliers_local_media_std)
    .sort_values("prop_outliers_local", ascending=False)
    .reset_index(drop=True)
)

print(outliers_local_media_std.head(48))
```

```python
# Media y std por barrio
mean_barrio = df_tp_clean.groupby("l3")["price_usd"].transform("mean")
std_barrio  = df_tp_clean.groupby("l3")["price_usd"].transform("std")

# Evitar división por cero (si algún barrio tiene std=0)
std_safe = std_barrio.replace(0, np.nan)

# z-score local por barrio
z_barrio = (df_tp_clean["price_usd"] - mean_barrio) / std_safe

# Si querés guardarlo en el df
df_tp_clean["z_price_barrio"] = z_barrio

# Flag de outlier local por z-score
df_tp_clean["outlier_z_local"] = (z_barrio.abs() > 2).astype(int)

# Proporción de outliers z por barrio (exploratorio)
prop_outliers_z = (
    df_tp_clean.groupby("l3")["outlier_z_local"]
    .mean()
    .reset_index()
    .sort_values("outlier_z_local", ascending=False)
)

print(prop_outliers_z.head(48))
```

# Integración de palabras claves al modelo

```python
#read csv df_modelado
df_modelado = pd.read_csv("/content/df_modelado (1).csv")
```

```python
df_modelado.shape
```

```python
df_modelado["title"] = df_modelado["title"].str.lower()
df_modelado["description"] = df_modelado["description"].str.lower()
```

```python
#crear columnas dummie de la palabra clave cuando title o description tengan alguna de las palabras clave
for palabra in palabras_clave:
    df_modelado[palabra] = (
        df_modelado["title"].str.contains(palabra, case=False, na=False) |
        df_modelado["description"].str.contains(palabra, case=False, na=False)
    ).astype(int)
```

```python
df_modelado.columns
```

```python
x = df_modelado[df_modelado.columns.drop("price_usd")]
y = df_modelado["price_usd"]
```

#Resultados finales

```python
scores = sk.model_selection.cross_val_score(rf, x, y, cv=10, scoring='neg_root_mean_squared_error', n_jobs=-1)

print(f"{scores=}")
print(-scores.mean())
print(scores.std())
```
